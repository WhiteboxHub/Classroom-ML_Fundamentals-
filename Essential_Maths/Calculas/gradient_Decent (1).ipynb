```python
import numpy as np

# Example dataset
X = np.array([1, 2, 3, 4, 5])  # Features
y = np.array([2.2, 4.0, 5.8, 8.5, 10.3])  # Labels

# Parameters to optimize
w = 0.0  # Weight
b = 0.0  # Bias

# Hyperparameters
learning_rate = 0.01
num_iterations = 1000

# Number of samples
m = len(X)

# Gradient descent loop
for i in range(num_iterations):
    # Predictions
    y_pred = w * X + b

    # Compute gradients
    dw = -(2/m) * np.sum((y - y_pred) * X)  # Gradient w.r.t. w
    db = -(2/m) * np.sum(y - y_pred)        # Gradient w.r.t. b

    # Update parameters
    w -= learning_rate * dw
    b -= learning_rate * db

    # Optional: Print loss every 100 iterations
    if i % 100 == 0:
        loss = np.mean((y - y_pred) ** 2)  # Mean squared error
        print(f"Iteration {i}: Loss = {loss:.4f}, w = {w:.4f}, b = {b:.4f}")

# Final parameters
print(f"Final parameters: w = {w:.4f}, b = {b:.4f}")

# Test the model
y_test_pred = w * X + b
print("Predicted values:", y_test_pred)
print("Actual values:   ", y)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Example dataset
X = np.array([1, 2, 3, 4, 5])  # Features
y = np.array([2.2, 4.0, 5.8, 8.5, 10.3])  # Labels

# Parameters to optimize
w = 0.0  # Weight
b = 0.0  # Bias

# Hyperparameters
learning_rate = 0.01
num_iterations = 1000

# Number of samples
m = len(X)

# To store loss values for plotting
loss_history = []

# Gradient descent loop
for i in range(num_iterations):
    # Predictions
    y_pred = w * X + b

    # Compute gradients
    dw = -(2/m) * np.sum((y - y_pred) * X)  # Gradient w.r.t. w
    db = -(2/m) * np.sum(y - y_pred)        # Gradient w.r.t. b

    # Update parameters
    w -= learning_rate * dw
    b -= learning_rate * db

    # Compute and store loss
    loss = np.mean((y - y_pred) ** 2)  # Mean squared error
    loss_history.append(loss)

    # Optional: Print loss every 100 iterations
    if i % 100 == 0:
        print(f"Iteration {i}: Loss = {loss:.4f}, w = {w:.4f}, b = {b:.4f}")

# Final parameters
print(f"Final parameters: w = {w:.4f}, b = {b:.4f}")

# Test the model
y_test_pred = w * X + b
print("Predicted values:", y_test_pred)
print("Actual values:   ", y)

# Plot the loss over iterations
plt.figure(figsize=(10, 6))
plt.plot(range(num_iterations), loss_history, label='Loss')
plt.xlabel('Iterations')
plt.ylabel('Mean Squared Error')
plt.title('Loss Curve')
plt.legend()
plt.grid()
plt.show()

# Plot the data and the fitted line
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', label='Data')
plt.plot(X, y_test_pred, color='red', label='Fitted Line')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression Fit')
plt.legend()
plt.grid()
plt.show()
```
