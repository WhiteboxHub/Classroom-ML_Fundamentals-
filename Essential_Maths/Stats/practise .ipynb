## Measure of spread or variability 

```python
data=pd.read_csv("/Users/innovapathinc/Desktop/Stats /house_dataset.csv")
```

```python
df=pd.DataFrame(data)
df.head()
```

### Range 
-Calculate Range: Difference between Max and Min

```python
range_sale_price = df['Sale Price'].max() - df['Sale Price'].min()
range_flat_area = df['Flat Area (in Sqft)'].max() - df['Flat Area (in Sqft)'].min()
print(range_sale_price)
print(range_flat_area)
```

### Variance 
Calculate Variance: Average of squared differences from the mean

```python
variance_sale_price = df['Sale Price'].var()
variance_flat_area = df['Flat Area (in Sqft)'].var()

print(variance_sale_price)
print(variance_flat_area)
```

### Standard Deviation 
calculate Standard Deviation: Square root of the variance


```python
std_dev_sale_price = df['Sale Price'].std()
std_dev_flat_area = df['Flat Area (in Sqft)'].std()
print(std_dev_sale_price)
print(std_dev_flat_area)
```

```python
# Display the results
print(f"Range of Sale Price: {range_sale_price}")
print(f"Range of Flat Area: {range_flat_area}")
print(f"Variance of Sale Price: {variance_sale_price}")
print(f"Variance of Flat Area: {variance_flat_area}")
print(f"Standard Deviation of Sale Price: {std_dev_sale_price}")
print(f"Standard Deviation of Flat Area: {std_dev_flat_area}")
```
