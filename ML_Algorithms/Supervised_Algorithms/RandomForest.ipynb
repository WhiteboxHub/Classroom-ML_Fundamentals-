```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv('heart_disease.csv')
df.isnull().sum()
```

```python
ip = df.drop('target',axis=1)
op = df['target']
```

```python
ip_train, ip_test , op_train, op_test = train_test_split(ip,op,random_state=42)
```

```python
random_forest = RandomForestClassifier(n_estimators=100,random_state=42)
random_forest.fit(ip_train,op_train)
```

```python
import pickle
with open('heart_disease_randomForest.pkl',mode='wb') as f:
    pickle.dump(random_forest,f)
```

```python

with open('heart_disease_randomForest.pkl',mode='rb') as f:
    model_rad_for = pickle.load(f)
```

```python
pred_op  = model_rad_for.predict(ip_test)
print(pred_op)
print(op_test)
```

```python
acc_score = accuracy_score(op_test,pred_op)
print(acc_score*100)
```

```python
cls_report = classification_report(op_test,pred_op)
print(cls_report)
```
